```python
import json
```

```python
def load_data(filename):
    with open(filename,"r") as f:
        data = json.load(f)
    return data
d = load_data("raw_data.json")
print(d)
print(type(d))
```

```python
def data_cleaning(d):
    cleaned = []
    unique_user = set()
    #cleaning data
    string_to_num={"one" : 1,"two" : 2,"three" : 3,"four" : 4,"five" : 5}
    for user in d:
        user_rating = str(user["rating"]).strip().lower()
        if(user_rating in string_to_num):
            user["rating"] = string_to_num[user_rating]
            print(user["rating"])

    #replacing missing data with none 
        raw_age = user.get("age")
        if(raw_age == None):
            user["age"] = None
    
        #removing duplicates
        if(user["name"].strip() not in unique_user):
            unique_user.add(user["name"].strip())
            cleaned.append(user)
    return cleaned
```

```python

cd = data_cleaning(d)
print(cd)
```

```python
#data insight
def data_insight(d):
    # average rating
    rating = 0
    per = 0
    for user in d:
        rating += float(user["rating"])
    #percentege of poor rating <3
        if(float(user["rating"]) < 3):
            per += 1
    print(f"Average rating of a data is {rating/len(d)}")
    print(f"Average percentage of a data is {(per/len(d))*100}")
```

```python
data_insight(cd)
```

```python
# recommending system
def recommend(data):
    recommendations = []
    for user in data:
        recom={}
        recom["name"] = user["name"]
        if(float(user["rating"]) >= 4):
            recom["recommended"] = "Apple"
        else:
            recom["recommended"] = "Samsung"
        recommendations.append(recom)
    return recommendations
```

```python
r = recommend(cd)
print(r)
```
