# Feature Engineering (Adding  and Removing columns)

```python
df2 = df.copy()
df2['tax'] = df2['income'].apply(lambda x : "20%" if x >60000 else "10%")
df2
assign_map = { "Male" : "M" , "Female" : "F" , "Unknown" : "U"}
df2['gender'] = df2['gender'].map(assign_map)
df2

df2.assign(new_income = df2['income'] * 1.1)

df2['country'] = df2['country'].replace('USA','US')
df2
```

```python
df2.columns = ['Id','Name','Age','Country','Gender','Income','Tax']
df2
df2.rename(columns = {'Income':'Salary'})
df2.rename(index = {1 : 'First'})

sorted_income = df2.sort_values(['Income'])
sorted_income = df2.sort_values(['Income'],ascending = False)
# print(sorted_income)

# df2.sort_index()
# sorted_income.reset_index()
# sorted_income.reset_index(drop = True)

sorted_income['Ranking'] = sorted_income['Income'].rank() # if two ranks are equal we will assing average
sorted_income['Ranking'] = sorted_income['Income'].rank(ascending = False)
sorted_income['Ranking'] = sorted_income['Income'].rank(method = 'dense') # removes float values in rank
sorted_income['Ranking'] = sorted_income['Income'].rank(method = 'min') # adds min values in rank
sorted_income['Ranking'] = sorted_income['Income'].rank(method = 'max') # adds max values in rank
sorted_income
sorted_income[['Id','Tax','Age','Country','Gender','Income','Name']]
sorted_income[['Name','Tax','Age','Country','Gender','Income','Id']]
```

# melt and pivot

```python
data = {
    'country' : ['India','us','India','us'],
    'year' : [2021,2020,2022,2023],
    'sales' : [23,234,43,4],
    'profit' : [100,20,30,30]
}
d = pd.DataFrame(data)
d
```

```python
melted = d.melt(
    id_vars = ['country','year'],
    value_vars = ['sales','profit'],
    var_name = 'metrix',
    value_name = 'value'
)
melted
```

```python
org = melted.pivot(
    index = ['country','year'],
    columns = ['metrix'],
    values = ['value']
)
org
```

# Merge and Joins

```python
order_data = pd.DataFrame({
    'name' : ['raj','laddu','charan','teja','surya'],
    'cust_id' : [1,2,3,4,5],
    'orders' : ['pen','pencil','eraser','box','egg']
})
customer_data =  pd.DataFrame({
    'cust_id' : [1,2,3,4,5],
    'name' : ['raj','laddu','charan','teja','surya']
})

pd.merge(order_data,customer_data,on = 'cust_id') # default inner join
pd.merge(order_data,customer_data,on = 'cust_id',how = 'outer') # outer join
pd.merge(order_data,customer_data,on = 'cust_id',how = 'left') #  left join
pd.merge(order_data,customer_data,on = 'cust_id',how = 'right') #  right join
pd.merge(order_data,customer_data,on = 'cust_id',how = 'inner') #  inner join
```
